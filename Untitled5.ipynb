```python
#Cell#2
def sigmoid(z):
  n=np.clip(z,-500,500)
  n=1/(1+np.exp(-n))
  return n
# Hint: use np.clip(z, -500, 500) to avoid overflow
def sigmoid_derivative(a):
  return a*(1-a)
# Hint: you derived this in Task 3 - express it in terms of a
```

```python
#Cell#3
def forward_pass(X, W1, b1, W2, b2, W3, b3):

# Layer 1
Z1 = X @ W1 + b1
 # (weighted sum: X @ W1 + b1)
A1 = sigmoid(Z1)
#  (sigmoid activation)
# Layer 2
Z2 = A1 @ W2 + b2
A2 = sigmoid(Z2)
# Output layer
Z3 = A2 @ W3 + b3
A3 = sigmoid(Z3)
return Z1, A1, Z2, A2, Z3, A3
```
